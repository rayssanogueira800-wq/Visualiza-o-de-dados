# visualização de dados 

Resumo: É transformar números em visuais para enteder padrões,tendencias e comunicar informações com clareza. O storytelling é contar uma história com os dados,dando sentido e mensagem ao que os números mostram

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

# matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np 

x = np.arange(1 , 6)
y = [8, 15, 6, 22, 18]

plt.figure(figsize=(6,3))
plt.bar(x,y, color='steelblue')
plt.xlabel("item")
plt.ylabel("quantidade")
plt.grid(axis='y',linestyle='--',alpha=0.5)
plt.show()
```
